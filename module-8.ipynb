In the first cell, we are simply creating data x and y to use in the provided tasks.

```python
# We need data to work with, lets create data first!
import numpy as np
x = np.arange(0,10)
y = x*2
z = x**2
```

```python
import matplotlib.pyplot as plt
# plt.show() for non-notebook users
```

**1. Instructions:**
* ** Use <code>plt.figure()</code> and create a figure object, say fig.**
* ** Add an axis to the figure canvas (`fig`), `[left, bottom, width, height] is [0,0,1,1]`. <i>Hint: <code>add_axes()</code><i>**
* ** Plot the data x,y on the axes**
* ** Set the labels x and y**
* ** Set the title**
* ** Create legends as shown in the plot**

Match the output below: 

```python
%matplotlib inline
import matplotlib.pyplot as plt
fig = plt.figure()
fig_axis = fig.add_axes([0,0,1,1])
fig_axis.plot(x,y,color='green',label='y = x*2')
fig_axis.set_xlabel('x_label')
fig_axis.set_ylabel('y_label')
fig_axis.set_title('Plot_title')
fig_axis.legend(loc='lower right')
plt.show()
```

**2. Instructions**<br>
* ** Create a figure object**
* ** Put two axes on the figure at [0,0,1,1] and [0.1,0.6,0.2,0.2]**
* ** Plot data (x,z) on main axes**
* ** Plot data (x,z) on the small axes and set its x limit b/w 1, 2, y limit 0, 5**
* ** Replicate the figure to axes, titles, legends etc**

```python
fig = plt.figure(figsize=(5, 3))
first = fig.add_axes([0, 0, 1, 1])
second = fig.add_axes([0.1, 0.6, 0.2, 0.2])

first.plot(x, z, color='red', label='y_main = x_main**2')
first.set_title('Title_main')
first.set_xlabel('x_main')
first.set_ylabel('y_main')
first.legend(loc='lower right')

second.plot(x, z, color='blue')
second.set_title('inset', fontsize=10)
second.set_xlabel('x_inset')
second.set_ylabel('y_inset')
second.set_xlim(1, 2)
second.set_ylim(0, 5)

plt.show()
```

**3. Replicate the output using <code>subplots()</code> and for loop** -- Note, the data is x and y. 

```python
plt.figure(figsize=(5,3))
for i in range(1,4):
    plt.subplot(1, 3, i)
    plt.plot(x,y)
plt.tight_layout()
plt.show()
```

**4. Replicate the output using <code>subplots()</code>**

**5. plot `(x,y)`, `(x,z)` and `(x,y)` on the above empty canvas to replicate the plot given below.**

```python
fig, (first, second, third) = plt.subplots(3, 1, figsize=(6, 4))
first.plot(x, y, 'b--')
second.plot(x, z, 'r-')
third.plot(x, y, 'g-')

plt.tight_layout()
plt.show()
```

**6. Resize the above plots to `figsize = (3,4)`**

```python
fig, (first, second, third) = plt.subplots(3, 1, figsize=(3, 4))
first.plot(x, y, 'b--')
second.plot(x, z, 'r-')
third.plot(x, y, 'g-')

plt.tight_layout()
plt.show()
```
